```python
def create_soc_alert(row, alert_id):
    return {
        "alert_id": alert_id,

        "service": {
            "destination_port": row["Destination Port"]
        },

        "flow": {
            "duration": row["Flow Duration"],
            "forward_packets": row["Total Fwd Packets"],
            "backward_packets": row["Total Backward Packets"],
            "forward_bytes": row["Total Length of Fwd Packets"],
            "backward_bytes": row["Total Length of Bwd Packets"]
        },

        "traffic": {
            "bytes_per_second": row["Flow Bytes/s"],
            "packets_per_second": row["Flow Packets/s"],
            "forward_packets_per_second": row["Fwd Packets/s"],
            "backward_packets_per_second": row["Bwd Packets/s"]
        },

        "packet_characteristics": {
            "forward_length_mean": row["Fwd Packet Length Mean"],
            "backward_length_mean": row["Bwd Packet Length Mean"],
            "packet_length_mean": row["Packet Length Mean"],
            "packet_length_std": row["Packet Length Std"]
        },

        "timing": {
            "flow_iat_mean": row["Flow IAT Mean"],
            "flow_iat_std": row["Flow IAT Std"],
            "flow_iat_max": row["Flow IAT Max"],
            "flow_iat_min": row["Flow IAT Min"]
        },

        "tcp_flags": {
            "fin": row["FIN Flag Count"],
            "syn": row["SYN Flag Count"],
            "rst": row["RST Flag Count"],
            "psh": row["PSH Flag Count"],
            "ack": row["ACK Flag Count"],
            "urg": row["URG Flag Count"]
        },

        "connection": {
            "init_window_forward": row["Init_Win_bytes_forward"],
            "init_window_backward": row["Init_Win_bytes_backward"],
            "active_mean": row["Active Mean"],
            "idle_mean": row["Idle Mean"]
        },

        "raw_features": row.to_dict()
    }
```

```python
def extract_attack_patterns(attack_data):
    documents = []

    for obj in attack_data["objects"]:

        if obj.get("type") != "attack-pattern":
            continue

        if obj.get("revoked") or obj.get("x_mitre_deprecated"):
            continue

        technique_name = obj.get("name", "")
        description = obj.get("description", "")

        technique_id = None

        for reference in obj.get("external_references", []):
            if reference.get("source_name") == "mitre-attack":
                technique_id = reference.get("external_id")
                break

        tactics = []

        for phase in obj.get("kill_chain_phases", []):
            if phase.get("kill_chain_name") == "mitre-attack":
                tactics.append(phase.get("phase_name"))

        platforms = obj.get("x_mitre_platforms", [])

        document = {
            "source": "MITRE ATT&CK",
            "type": "attack-pattern",
            "name": technique_name,
            "technique_id": technique_id,
            "description": description,
            "tactics": tactics,
            "platforms": platforms,
            "is_subtechnique": obj.get("x_mitre_is_subtechnique", False)
        }

        documents.append(document)

    return documents
```

```python
embedding_model = SentenceTransformer("all-MiniLM-L6-v2")

print("Embedding model loaded successfully.")
```

```python
print("Embedding dimension:", embedding_model.get_sentence_embedding_dimension())
```

```python
mitre_texts = [
    document["text"]
    for document in mitre_documents
]

mitre_embeddings = embedding_model.encode(
    mitre_texts,
    show_progress_bar=True
)

print("Number of embeddings:", len(mitre_embeddings))
print("Embedding shape:", mitre_embeddings.shape)
```

```python
import faiss
import numpy as np

# Convert embeddings to the format FAISS expects
mitre_embeddings = np.asarray(mitre_embeddings).astype("float32")

# Create a FAISS index using Euclidean distance
dimension = mitre_embeddings.shape[1]

index = faiss.IndexFlatL2(dimension)

# Add all MITRE embeddings
index.add(mitre_embeddings)

print("FAISS index created successfully.")
print("Number of vectors:", index.ntotal)
print("Vector dimension:", dimension)
```

```python
def retrieve_mitre_evidence(query, top_k=5):
    query_embedding = embedding_model.encode(
        [query]
    ).astype("float32")

    distances, indices = index.search(
        query_embedding,
        top_k
    )

    results = []

    for distance, idx in zip(distances[0], indices[0]):
        document = mitre_documents[idx]

        results.append({
            "technique_id": document["technique_id"],
            "name": document["name"],
            "distance": float(distance),
            "description": document["description"],
            "tactics": document["tactics"],
            "platforms": document["platforms"]
        })

    return results
```

```python
def extract_cve_document(cve_item):
    # English description
    description = ""

    for item in cve_item.get("descriptions", []):
        if item.get("lang") == "en":
            description = item.get("value", "")
            break

    # CWE
    weaknesses = []

    for weakness in cve_item.get("weaknesses", []):
        for item in weakness.get("description", []):
            if item.get("lang") == "en":
                weaknesses.append(item.get("value", ""))

    # Affected products
    affected_products = []

    for affected in cve_item.get("affected", []):
        vendor = affected.get("vendor", "unknown")
        product = affected.get("product", "unknown")

        affected_products.append({
            "vendor": vendor,
            "product": product
        })

    # CVSS information
    cvss = {}

    metrics = cve_item.get("metrics", {})

    for metric_type, metric_list in metrics.items():

        if not metric_list:
            continue

        metric = metric_list[0]

        cvss_data = metric.get("cvssData", {})

        cvss[metric_type] = {
            "version": cvss_data.get("version"),
            "base_score": cvss_data.get("baseScore"),
            "vector": cvss_data.get("vectorString")
        }

    document = {
        "source": "NVD",
        "type": "cve",
        "cve_id": cve_item.get("id"),
        "description": description,
        "published": cve_item.get("published"),
        "last_modified": cve_item.get("lastModified"),
        "status": cve_item.get("vulnStatus"),
        "weaknesses": weaknesses,
        "affected_products": affected_products,
        "cvss": cvss
    }

    return document
```

```python
def create_cve_text(document):
    weaknesses = ", ".join(document["weaknesses"])

    products = []

    for product in document["affected_products"]:
        products.append(
            f'{product["vendor"]} {product["product"]}'
        )

    products_text = ", ".join(products)

    cvss_parts = []

    for metric_type, metric in document["cvss"].items():
        cvss_parts.append(
            f'{metric_type}: '
            f'CVSS {metric["version"]}, '
            f'base score {metric["base_score"]}, '
            f'vector {metric["vector"]}'
        )

    cvss_text = "; ".join(cvss_parts)

    text = f"""
Source: {document["source"]}
Type: {document["type"]}
CVE ID: {document["cve_id"]}

Description:
{document["description"]}

Published:
{document["published"]}

Last Modified:
{document["last_modified"]}

Status:
{document["status"]}

Weaknesses:
{weaknesses}

Affected Products:
{products_text}

CVSS:
{cvss_text}
""".strip()

    return text
```

```python
def retrieve_nvd_evidence(query, top_k=5):
    query_embedding = embedding_model.encode(
        [query]
    ).astype("float32")

    distances, indices = nvd_index.search(
        query_embedding,
        top_k
    )

    results = []

    for distance, idx in zip(distances[0], indices[0]):
        document = nvd_documents[idx]

        results.append({
            "source": "NVD",
            "cve_id": document["cve_id"],
            "distance": float(distance),
            "description": document["description"],
            "weaknesses": document["weaknesses"],
            "affected_products": document["affected_products"],
            "cvss": document["cvss"]
        })

    return results
```

```python

def build_retrieval_plan(alert):
    service = alert.get("service", {})
    flow = alert.get("flow", {})
    traffic = alert.get("traffic", {})
    flags = alert.get("tcp_flags", {})
    timing = alert.get("timing", {})

    # Summarize only facts actually present in the alert.
    observed_facts = [
        f"Destination port: {service.get('destination_port')}",
        f"Flow duration: {flow.get('duration')}",
        f"Forward packets: {flow.get('forward_packets')}",
        f"Backward packets: {flow.get('backward_packets')}",
        f"Forward bytes: {flow.get('forward_bytes')}",
        f"Backward bytes: {flow.get('backward_bytes')}",
        f"Bytes per second: {traffic.get('bytes_per_second')}",
        f"Packets per second: {traffic.get('packets_per_second')}",
        f"TCP flags: {flags}",
        f"Flow inter-arrival timing: {timing}"
    ]

    # Broad exploratory query; this does not establish an attack.
    mitre_query = (
        "Enterprise cybersecurity knowledge about interpreting network "
        "traffic and connection-level telemetry. The observations include "
        "flow duration, packet and byte counts, traffic rates, TCP flags, "
        "and packet timing. Identify relevant techniques only when their "
        "documented behavior is supported by additional evidence. "
        "Do not assume a specific attack technique."
    )

    # Search NVD only if the alert contains relevant asset/vulnerability context.
    context_keys = [
        "software",
        "software_name",
        "software_version",
        "cve_id",
        "vulnerability_id",
        "affected_product"
    ]

    has_asset_context = any(
        alert.get(key) for key in context_keys
    )

    return {
        "alert_id": alert.get("alert_id"),
        "observed_facts": observed_facts,
        "mitre_query": mitre_query,
        "search_nvd": bool(has_asset_context),
        "nvd_reason": (
            "Asset or vulnerability context is available."
            if has_asset_context
            else (
                "Skipped: no software/version, CVE ID, or affected-product "
                "context is present in this alert."
            )
        ),
        "applicability_rule": (
            "Retrieved documents are candidates, not proof that a "
            "technique or vulnerability applies to this alert."
        )
    }


retrieval_plan = build_retrieval_plan(sample_alert)

print("ALERT:", retrieval_plan["alert_id"])

print("\nOBSERVED FACTS")
for fact in retrieval_plan["observed_facts"]:
    print("-", fact)

print("\nMITRE ATT&CK QUERY")
print(retrieval_plan["mitre_query"])

print("\nNVD SEARCH:", retrieval_plan["search_nvd"])
print(retrieval_plan["nvd_reason"])

print("\nAPPLICABILITY RULE")
print(retrieval_plan["applicability_rule"])
```

```python

from pathlib import Path
import json
import numpy as np
import faiss
from sentence_transformers import SentenceTransformer

project_dir = Path(
    "/home/jupyter-itproject2026/TrustSOC"
)

documents_path = (
    project_dir / "data/knowledge_base/mitre_attack/"
    "mitre_attack_documents.json"
)

# Load the saved MITRE ATT&CK documents.
with open(documents_path, "r", encoding="utf-8") as file:
    mitre_documents = json.load(file)

print("MITRE documents loaded:", len(mitre_documents))

# Reuse the model if it exists; otherwise load its cached/downloaded copy.
if "embedding_model" not in globals():
    embedding_model = SentenceTransformer("all-MiniLM-L6-v2")

# Rebuild the vector index from the saved documents.
mitre_texts = [
    document.get("text", "")
    for document in mitre_documents
]

mitre_embeddings = embedding_model.encode(
    mitre_texts,
    show_progress_bar=True,
    convert_to_numpy=True
)

mitre_embeddings = np.asarray(
    mitre_embeddings, dtype="float32"
)

index = faiss.IndexFlatL2(mitre_embeddings.shape[1])
index.add(mitre_embeddings)


def retrieve_mitre_evidence(query, top_k=5):
    query_embedding = embedding_model.encode(
        [query],
        convert_to_numpy=True
    )
    query_embedding = np.asarray(
        query_embedding, dtype="float32"
    )

    distances, indices = index.search(
        query_embedding,
        min(top_k, len(mitre_documents))
    )

    results = []

    for distance, document_index in zip(
        distances[0], indices[0]
    ):
        if document_index < 0:
            continue

        document = mitre_documents[int(document_index)]

        results.append({
            "source": document.get("source", "MITRE ATT&CK"),
            "technique_id": document.get("technique_id"),
            "name": document.get("name"),
            "distance": float(distance),
            "description": document.get("description", ""),
            "tactics": document.get("tactics", []),
            "platforms": document.get("platforms", []),
            "is_subtechnique": document.get(
                "is_subtechnique", False
            )
        })

    return results


print("FAISS vectors:", index.ntotal)
print("Embedding dimensions:", index.d)
print("MITRE retriever restored successfully.")
```

```python

import json
import re

def explain_threat_assessment(assessment):
    allowed_input = {
        "alert_id": assessment["alert_id"],
        "overall_assessment": assessment["overall_assessment"],
        "observed_telemetry": assessment["observed_telemetry"],
        "missing_context": assessment["missing_context"],
        "mitre_candidates": [
            {
                "technique_id": item["technique_id"],
                "technique_name": item["technique_name"],
                "applicability": item["applicability"],
                "reason": item["reason"]
            }
            for item in assessment["mitre_candidates"]
        ],
        "analyst_guidance": assessment["analyst_guidance"]
    }

    prompt = f"""
You are explaining a structured cybersecurity assessment to a SOC analyst.

Use only the information in the JSON below.
Do not add IP addresses, hostnames, users, software, CVEs, attack claims,
or technical observations that are not explicitly present.
Do not change the assessment or claim that a candidate technique occurred.
Explain that missing context limits conclusions.
Use 3 to 5 short sentences.

STRUCTURED ASSESSMENT:
{json.dumps(allowed_input, ensure_ascii=False)}

Write the analyst explanation:
"""

    inputs = qwen_tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=2500
    ).to(device)

    with torch.no_grad():
        outputs = qwen_model.generate(
            **inputs,
            max_new_tokens=140,
            do_sample=False,
            pad_token_id=qwen_tokenizer.eos_token_id
        )

    generated_tokens = outputs[0][inputs["input_ids"].shape[1]:]
    explanation = qwen_tokenizer.decode(
        generated_tokens,
        skip_special_tokens=True
    ).strip()

    if not explanation:
        explanation = (
            "The assessment is inconclusive. Network-flow telemetry is "
            "available, but endpoint, identity, asset, and security-event "
            "context are missing. The retrieved MITRE ATT&CK techniques "
            "remain candidates rather than established findings."
        )

    return explanation


explanation = explain_threat_assessment(threat_intel_assessment)

print("QWEN ANALYST EXPLANATION")
print("=" * 60)
print(explanation)
```

```python

def validate_qwen_explanation(explanation, assessment):
    text = explanation.strip()

    # Reject empty or visibly incomplete output.
    if not text or text[-1] not in ".!?":
        return False, "Output is empty or appears incomplete."

    # Reject common unsupported incident claims.
    unsupported_phrases = [
        "the attacker",
        "the attacker's",
        "extent of damage",
        "damage caused",
        "exploited by the attacker",
        "severity of the incident",
        "the vulnerability was exploited"
    ]

    lowered = text.lower()

    for phrase in unsupported_phrases:
        if phrase in lowered:
            return False, f"Potentially unsupported claim: {phrase}"

    return True, "Basic checks passed."


def create_deterministic_explanation(assessment):
    available = [
        category.replace("_", " ")
        for category, fields in assessment["observed_telemetry"].items()
        if fields
    ]

    missing = [
        category.replace("_", " ")
        for category in assessment["missing_context"]
    ]

    candidate_count = len(assessment["mitre_candidates"])

    available_text = ", ".join(available) if available else "none"
    missing_text = ", ".join(missing) if missing else "none"

    return (
        f'Alert {assessment["alert_id"]} has an overall assessment of '
        f'{assessment["overall_assessment"]}. Available telemetry includes '
        f'{available_text}. Missing context includes {missing_text}. '
        f'{candidate_count} MITRE ATT&CK candidates were retrieved, but '
        f'none is established as applicable by this assessment. '
        f'Insufficient information does not mean the alert is benign.'
    )


is_valid, validation_reason = validate_qwen_explanation(
    explanation,
    threat_intel_assessment
)

if is_valid:
    final_explanation = explanation
    explanation_source = "Qwen (passed basic output checks)"
else:
    final_explanation = create_deterministic_explanation(
        threat_intel_assessment
    )
    explanation_source = "Deterministic Python fallback"

print("Explanation source:", explanation_source)
print("Validation:", validation_reason)
print("\nFINAL ANALYST EXPLANATION")
print("=" * 60)
print(final_explanation)
```

```python

def apply_evidence_gate(alert, retrieved_evidence):
    inventory = inspect_alert_evidence(alert)

    has_endpoint = bool(
        inventory["endpoint_telemetry"]["present"]
    )
    has_identity = bool(
        inventory["identity_telemetry"]["present"]
    )
    has_asset_context = bool(
        inventory["asset_and_vulnerability_context"]["present"]
    )

    gated_candidates = []

    for item in retrieved_evidence:
        name = item.get("name", "Unknown technique")

        gated_candidates.append({
            "technique_id": item.get("technique_id"),
            "name": name,
            "retrieval_distance": item.get("distance"),
            "retrieval_status": "candidate_retrieved",
            "applicability": "insufficient_information",
            "investigation_priority": "unassessed",
            "reason": (
                "Semantic retrieval returned this candidate, but the "
                "current alert does not establish the behavior described "
                "by the technique."
            ),
            "available_context": {
                "network_flow": True,
                "endpoint_telemetry": has_endpoint,
                "identity_telemetry": has_identity,
                "asset_context": has_asset_context
            }
        })

    return {
        "alert_id": alert.get("alert_id"),
        "overall_assessment": "insufficient_information",
        "candidates": gated_candidates,
        "note": (
            "No candidate is confirmed by semantic retrieval alone. "
            "Investigation priority has not been assessed."
        )
    }


# Test the gate with the hybrid query results for ALERT_0001.
alert_1_results = retrieve_mitre_evidence(
    build_hybrid_retrieval_query(alert_1),
    top_k=5
)

gated_report = apply_evidence_gate(alert_1, alert_1_results)

print("Alert:", gated_report["alert_id"])
print("Assessment:", gated_report["overall_assessment"])

for candidate in gated_report["candidates"]:
    print(
        f'\n{candidate["technique_id"]} — {candidate["name"]}'
    )
    print("Retrieval:", candidate["retrieval_status"])
    print("Applicability:", candidate["applicability"])
    print("Investigation priority:", candidate["investigation_priority"])
    print("Reason:", candidate["reason"])
```
